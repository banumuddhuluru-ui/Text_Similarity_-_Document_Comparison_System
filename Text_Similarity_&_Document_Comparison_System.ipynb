```python
# ============================================================
# NLP-BASED TEXT SIMILARITY ANALYZER
# Using TF-IDF + Cosine Similarity + Gradio
# ============================================================

!pip install -q gradio scikit-learn

import gradio as gr
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ------------------------------------------------------------
# TEXT SIMILARITY FUNCTION
# ------------------------------------------------------------

def compare_texts(text1, text2):

    if not text1.strip() or not text2.strip():
        return (
            "Please enter both texts.",
            "0%",
            "Both text fields are required.",
            "Waiting for input..."
        )

    # Create TF-IDF vectors
    vectorizer = TfidfVectorizer()

    vectors = vectorizer.fit_transform([text1, text2])

    # Calculate cosine similarity
    similarity = cosine_similarity(vectors[0:1], vectors[1:2])[0][0]

    percentage = similarity * 100

    # Analysis
    if percentage >= 80:
        analysis = "The two texts are highly similar."

    elif percentage >= 50:
        analysis = "The two texts have moderate similarity."

    elif percentage >= 20:
        analysis = "The two texts have low similarity."

    else:
        analysis = "The two texts are mostly different."

    result = "Text comparison completed successfully."

    return (
        f"{percentage:.2f}%",
        f"{percentage:.2f}%",
        analysis,
        result
    )


# ------------------------------------------------------------
# CUSTOM CSS
# ------------------------------------------------------------

css = """
body {
    background-color: #0b0f19 !important;
}

.gradio-container {
    background-color: #0b0f19 !important;
    color: white !important;
    max-width: 1200px !important;
}

h1 {
    color: white !important;
    font-size: 34px !important;
}

h2, h3, p, label {
    color: white !important;
}

textarea, input {
    background-color: #374151 !important;
    color: white !important;
}

button {
    background-color: #5146e5 !important;
    color: white !important;
    font-weight: bold !important;
}
"""


# ------------------------------------------------------------
# WEB APPLICATION
# ------------------------------------------------------------

with gr.Blocks(
    theme=gr.themes.Base(),
    css=css,
    title="NLP-Based Text Similarity Analyzer"
) as demo:

    gr.Markdown(
        """
        # 📄 NLP-Based Text Similarity Analyzer

        ### Compare two texts using Natural Language Processing

        This application uses **TF-IDF and Cosine Similarity**
        to measure how similar two pieces of text are.
        """
    )

    gr.Markdown(
        """
        **Features:**

        - 📝 Compare two texts
        - 📊 Similarity percentage
        - 🔎 Text analysis
        - 🧠 TF-IDF based comparison
        """
    )

    with gr.Row():

        # TEXT 1
        with gr.Column():

            gr.Markdown("### 📄 Text 1")

            text1 = gr.Textbox(
                label="First Text",
                placeholder="Enter the first text here...",
                lines=8
            )

        # TEXT 2
        with gr.Column():

            gr.Markdown("### 📄 Text 2")

            text2 = gr.Textbox(
                label="Second Text",
                placeholder="Enter the second text here...",
                lines=8
            )

    compare_button = gr.Button(
        "🔍 Compare Texts",
        variant="primary"
    )

    gr.Markdown("### 📊 Comparison Result")

    with gr.Row():

        similarity_output = gr.Textbox(
            label="Similarity Percentage",
            interactive=False
        )

        score_output = gr.Textbox(
            label="Similarity Score",
            interactive=False
        )

    analysis_output = gr.Textbox(
        label="Analysis",
        lines=3,
        interactive=False
    )

    result_output = gr.Textbox(
        label="Result",
        interactive=False
    )

    # Example inputs
    gr.Markdown("### 💡 Example")

    gr.Examples(
        examples=[
            [
                "Artificial intelligence helps computers learn from data.",
                "Machine learning allows computers to learn from data."
            ],
            [
                "Python is a programming language.",
                "Python is widely used for software development."
            ]
        ],
        inputs=[text1, text2]
    )

    # Button action
    compare_button.click(
        fn=compare_texts,
        inputs=[text1, text2],
        outputs=[
            similarity_output,
            score_output,
            analysis_output,
            result_output
        ]
    )


# ------------------------------------------------------------
# LAUNCH
# ------------------------------------------------------------

demo.launch(share=True)
```
